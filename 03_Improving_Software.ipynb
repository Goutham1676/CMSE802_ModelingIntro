# Exercise 2: Commit Your Current Work

Before using automated tools to modify your code, create a Git commit that captures the current state of your project.

Use a meaningful commit message that describes the work you have completed.

As a general rule, avoid:

```bash
git add *
```

and

```bash
git add .
```

These commands often add files that you did not intend to commit such as hidden files.

Instead, be intentional about the files you add.

If you want to stage modifications to files that Git is already tracking, consider using:

```bash
git add -u
```

You can then add any new files individually.

Example:

```bash
git add -u
git add barycentric_triangle.py
git commit -m "Refactored triangle script into reusable library"
```

Avoid vague commit messages such as:

```text
update
stuff
changes
fixes
```

A good commit message should help both you and future collaborators understand what changed and why.

Remember that Git is not just a backup system. It is a record of how a project evolves over time.

# Exercise 7: Format the Code

Now run:

```bash
ruff format .
```

This command focuses on code formatting and readability.

Before looking at your editor, ask Git what changed:

```bash
git diff
```

Git will show the differences between your current files and the last committed version.

Spend a few minutes scrolling through the output and reviewing the modifications Ruff made.

Remember that formatting tools are designed to improve consistency, but it is still your responsibility to understand changes being made to your code.
